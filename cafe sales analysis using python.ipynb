# Dtype conversion 


```python

df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce') 
text_cols = ['Item', 'Payment Method', 'Dining Option']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()

df['Quantity'] = pd.to_numeric(df['Quantity'].astype(str).str.strip(), errors='coerce')
for col in ['Price Per Unit', 'Total Spent']:
    # Clean out currency signs ($), commas, and spaces first
    df[col] = df[col].astype(str).str.replace('$', '', regex=False).str.strip()
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df.dtypes)
```

# missing values treatment in payment category column

```python
# Step 1: Clean and standardize text to Title Case
df['Payment Method'] = df['Payment Method'].astype(str).str.strip().str.title()

# Step 2: Replace invalid placeholders with 'Unspecified'
invalid_values = ['Error', 'Unknown', '0', 'Nan', 'None', '']
df['Payment Method'] = df['Payment Method'].replace(invalid_values, 'Unspecified')

# Step 3: Handle any remaining actual NaN values
df['Payment Method'] = df['Payment Method'].fillna('Unspecified')

df
```

```python
df.isnull().sum().sum()
```

```python
print(df.isna().sum())
```

# working on Transaction date column 

```python
# Forward fill missing dates from preceding transactions
df['Transaction Date'] = df['Transaction Date'].ffill()

# Recalculate derived month and weekday
df['Transaction Month'] = df['Transaction Date'].dt.month_name()
df['Transaction weekday'] = df['Transaction Date'].dt.day_name()
```

```python
# Drop rows where Transaction Date is NaT
df = df.dropna(subset=['Transaction Date'])

# Re-derive month and weekday
df['Transaction Month'] = df['Transaction Date'].dt.month_name()
df['Transaction weekday'] = df['Transaction Date'].dt.day_name()
```

```python
print(df.isna().sum())
```

# working on numerical columns missing values 

```python
# replacing records tht have unknown , error, 0 from payment status 

import numpy as np

# Step 1: Use exact row-level math where 2 out of 3 numbers exist
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent'] / df['Quantity'])
df['Quantity'] = df['Quantity'].fillna(df['Total Spent'] / df['Price Per Unit'])

# Step 2: Fill remaining missing Price Per Unit using median price of that specific Item
df['Price Per Unit'] = df['Price Per Unit'].fillna(
    df.groupby('Item')['Price Per Unit'].transform('median')
)

# Step 3: Re-apply row math after Price Per Unit is imputed
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
df['Quantity'] = df['Quantity'].fillna(df['Total Spent'] / df['Price Per Unit'])

# Step 4: Final fallback for remaining Quantity (default to 1) and calculate final Total Spent
df['Quantity'] = df['Quantity'].fillna(1)
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
```

# Categorical column DATA MANIPULATION

##  work on payment option column


```python
# cash as offline / digital wallet or credit card as online 
df['Payment Category'] = df['Payment Method']
df
```

```python
# List all categorical text columns
cat_cols = ['Item', 'Payment Method', 'Dining Option', 'Payment Category']

for col in cat_cols:
    # Convert to string and clean extra whitespace
    df[col] = df[col].astype(str).str.strip()
    
    # Replace string artifacts ('Nan', 'UNKNOWN', 'Unknown', '0', 'None', etc.) with 'Unspecified'
    df[col] = df[col].replace(
        ['Nan', 'nan', 'UNKNOWN', 'Unknown', 'ERROR', 'Error', '0', 'None', '', 'nan'], 
        'Unspecified'
    )

    # Handle true NaN values if any exist
    df[col] = df[col].fillna('Unspecified')
```

```python
df
```

```python
# Map Payment Methods to Offline / Online categories
payment_map = {
    'Cash': 'Offline',
    'Credit Card': 'Online',
    'Digital Wallet': 'Online',
    'Unspecified': 'Unspecified'
}

df['Payment Category'] = df['Payment Method'].map(payment_map).fillna('Unspecified')
```

```python
df
```

```python
# categorise food items as eatables and drinks 
print(df['Item'].unique())
```

```python
# 1. Define the item categorization map
item_category_map = {
    'Coffee': 'Drink',
    'Tea': 'Drink',
    'Juice': 'Drink',
    'Smoothie': 'Drink',
    'Cake': 'Eatable',
    'Cookie': 'Eatable',
    'Sandwich': 'Eatable',
    'Salad': 'Eatable',
    'Unspecified': 'Unspecified'
}

# 2. Create the new 'Item Category' column
df['Item Category'] = df['Item'].map(item_category_map).fillna('Unspecified')

df
```

```python
# Create a new column showing total sales amount per Transaction ID
df['ID Total Sales'] = df.groupby('Transaction ID')['Total Spent'].transform('sum')
```
